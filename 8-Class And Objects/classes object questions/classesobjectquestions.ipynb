```python
class BankAccount:
    def __init__(self,account_number,balance):
        self.account_number = account_number
        self.balance = balance
    def deposit(self,money):
        self.balance += money
        return print(f"{money} Deposited in your account. New balance is {self.balance}")
    def withdraw(self,money):
        self.balance -= money
        return print(f"{money} Withdrwed in your account. New balance is {self.balance}")

hooman_acc = BankAccount(1234,100)
hooman_acc.deposit(100)
hooman_acc.withdraw(50)

```
